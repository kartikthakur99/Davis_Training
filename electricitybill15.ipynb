```python
units = int(input("Enter units: "))

if units <= 50:
    bill = units * 3
elif units <= 100:
    bill = (50 * 3) + (units - 50) * 5
else:
    bill = (50 * 3) + (50 * 5) + (units - 100) * 7

print("Total bill amount:", bill)
```
